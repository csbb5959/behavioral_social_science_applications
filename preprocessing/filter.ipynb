```python
import sqlite3
import pandas as pd
DATABASE_PATH = "data/podcastindex_feeds.db"
```

```python
con = sqlite3.connect("../data/podcastindex_feeds.db")
cur = con.cursor()
```

```python
cur.execute("SELECT name FROM sqlite_master")
cur.fetchall()
```

```python
cur.execute("PRAGMA table_info(podcasts)")
cur.fetchall()
```

```python
cur.execute("SELECT language FROM podcasts LIMIT 10")
cur.fetchall()
```

```python
cur.execute("SELECT id, url, language, category1, category2, category3, newestItemPubdate, oldestItemPubdate FROM podcasts WHERE language LIKE '%en' AND dead = 0 LIMIT 20")
cur.fetchall()
```

```python
cur.execute('SELECT * FROM podcasts WHERE 1=1 AND language IN (?,?) AND category1 IN (?,?) AND category2 IN (?,?) AND category3 IN (?,?) AND newestItemPubdate > (?) AND oldestItemPubdate < (?) AND dead IN (?) LIMIT 50', ['en', 'en-us', 'games', 'science', 'hobbies', 'medicine', 'video games', 'education', 2023, 2022, '0'])
cur.fetchall()
```

```python
cur.execute("SELECT DISTINCT category1, category2, category3 AS distinct_cat1, COUNT(category1) AS cnt_distinct_cat1 FROM podcasts WHERE language LIKE '%en' AND dead = 0 GROUP BY category1 ORDER BY cnt_distinct_cat1 DESC LIMIT 200")
cur.fetchall()
```
