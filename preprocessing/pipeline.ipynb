```python
from filter_index import PodcastDatabase
from extract_episodes import EpisodeFetcher
import pandas as pd

DATABASE_PATH = "../data/podcastindex_feeds.db"
TABLE = "podcasts"
YEARS = [2022, 2023, 2024, 2025, 2026]
filters = {"language": ["en", "en-us"], 
               "category1": ["games", "science"], 
               "category2": ["hobbies", "medicine"], 
               "category3": ["video games", "education"],
               "newestItemPubdate": [2026],
               "oldestItemPubdate": [2018],
               "dead": ["0"]}
k = 2
MIN_PODCAST_LEN_MINUTES = 15

pod_db = PodcastDatabase(DATABASE_PATH, TABLE)
ep_fetcher = EpisodeFetcher()
```

```python
feed_urls = pod_db.filter(filters)
pod_db.close()
feed_urls
```

```python
episodes_df = pd.DataFrame(ep_fetcher.fetch_episodes_for_podcasts(feed_urls))
episodes_df.head()
```

```python
len(episodes_df)
```

```python
episodes_df = episodes_df.dropna()
len(episodes_df)
```

```python
year_mask = episodes_df["episode_year"].isin(YEARS) # filter out episodes that are not in the desired years
episodes_df = episodes_df[year_mask]
episodes_df
```

```python
episodes_df = episodes_df.astype({"episode_length": "int32"})
length_mask = episodes_df["episode_length"] >= MIN_PODCAST_LEN_MINUTES * 60
episodes_df = episodes_df[length_mask]
episodes_df
```

```python
counts = (episodes_df.groupby(["feed_url", "episode_year"]).size()
            .unstack(fill_value=0)                 # Zeilen: feed_url, Spalten: Jahre
            .reindex(columns=YEARS, fill_value=0)) # fehlende Jahre explizit als 0
keep = counts.index[(counts >= k).all(axis=1)]
df = episodes_df[episodes_df["feed_url"].isin(keep)]
df
```

```python
feed_urls_unique = df["feed_url"].unique()
```

```python
import numpy as np
```

```python
rng = np.random.default_rng(23)
randomized_pds = rng.choice(feed_urls_unique, size=3, replace=False)
randomized_pds
```

```python
randomized_pd_df = df[df["feed_url"].isin(randomized_pds)]
randomized_pd_df
randomized_pd_df["feed_url"].unique()
```

```python
sampled_df = randomized_pd_df.groupby(['feed_url', "episode_year"]).sample(n=2, random_state=23)
sampled_df
```
