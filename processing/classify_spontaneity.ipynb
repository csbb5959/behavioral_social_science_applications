```python
import pandas as pd
import os, json
from groq import Groq
client = Groq(api_key=os.environ["GROQ_API_KEY"])
MODEL = "openai/gpt-oss-120b"
PROMPT = """You get an excerpt from a podcast transcript.
Rate how spontaneous the speech is:
1 = clearly scripted (read from a prepared text, like news or an audiobook)
2 = mostly scripted (prepared text with a few free parts)
3 = mostly spontaneous (free talk with a few prepared parts)
4 = clearly spontaneous (free conversation: hesitations, interruptions, self-corrections)
First give a one-sentence reason, then the score.
Answer only with JSON: {"reason": "...", "score": <1-4>}"""
SCORE_THRESHOLD = 3
```

```python
transcript_text = "Welcome back to Tech Synergy, the podcast where we break down complex digital transformation into bite-sized actionable insights. I’m your host, Marcus Vance. " \
"And I’m Elena Rostova. Today, Marcus, we are diving deep into a topic that every modern enterprise leader is talking about: artificial intelligence in workflow automation. That’s absolutely " \
"right, Elena. In fact, a recent report highlighted that organizations adopting automated workflows see a thirty percent increase in operational efficiency within the first quarter alone. That is a " \
"staggering figure, Marcus. But before our listeners rush out to implement these tools, we need to ask the crucial question: where should a company actually start? Great question, Elena. It all comes down " \
"to what industry experts call The Three Pillars of Integration: assessment, scalability, and security. Without a solid assessment phase, any software deployment is bound to hit unexpected roadblocks. Exactly. " \
"You can't fix a bottleneck if you haven't mapped out where the congestion occurs in the first place. Perfectly stated. Joining us today to unpack this framework is Dr. Aris Thorne, Chief Innovation Officer at DataPulse " \
"Solutions. Dr. Thorne, welcome to the show. Thanks for having me, Marcus and Elena. It’s a pleasure to be here to discuss how organizations can turn operational friction into competitive advantage. We’re thrilled to have you, " \
"Dr. Thorne. Let’s start with that first pillar: assessment. What is the single biggest mistake companies make when evaluating their current processes? We will get into that answer right after a quick word from our sponsor. Stay tuned."
```

```python
response = client.chat.completions.create(
                model=MODEL,
                temperature=0,
                response_format={"type": "json_object"},
                messages=[{"role": "system", "content": PROMPT},
                        {"role": "user", "content": transcript_text}],
            )
```
